## linear regression

### data creation

```python
WEIGHT=0.7
BIAS=0.3
```

```python
X=torch.arange(0,1,0.02).unsqueeze(1)
X
```

```python
X.shape ##batch or number of data points,features
```

```python
y=WEIGHT*X+BIAS
```

```python
y.shape
```

```python
train_size=int(0.8*(len(X)))
train_size
```

```python
X_train,y_train=X[:train_size],y[:train_size]
X_test,y_test=X[train_size:],y[train_size:]
```

```python
X_train.shape,y_train.shape
```

```python
X_test.shape,y_test.shape
```

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(X_train,y_train,s=25,c="g",marker="s")##s is size ## c controls color
plt.scatter(X_test,y_test)
plt.title("Linear Regression Data")
plt.show()
```

### Model building

```python
import torch.nn as nn ## or from torch import nn
```

```python
class LinearRegressionModel(nn.Module):
  def __init__(self):
    super().__init__()
    self.weights=nn.Parameter(torch.randn(1,dtype=torch.float32),requires_grad=True)
    self.bias=nn.Parameter(torch.randn(1,dtype=torch.float32),requires_grad=True)

  def forward(self,x:torch.Tensor)->torch.Tensor:
    return self.weights * x + self.bias
```

```python
model=LinearRegressionModel()
```

```python
model.state_dict()
```

```python
model.eval()##putting model into evaluation mode
with torch.inference_mode():
  y_pred_untrained=model(X_test)
  plt.scatter(X_test,y_test)
  plt.scatter(X_test,y_pred_untrained)
  plt.show()
```

```python
import torch.optim as optim
```

```python
loss_fn=nn.MSELoss()
optimizer=optim.SGD(model.parameters(),lr=0.01)
EPOCHS=200
train_loss_history=[]
test_loss_history=[]
for epoch in range(EPOCHS):
  model.train()##puts model in training mode
  y_pred=model(X_train)
  train_loss=loss_fn(y_pred,y_train)
  optimizer.zero_grad()
  train_loss.backward()## computing gradient
  optimizer.step() ## updating the parameters
  model.eval() ## putting in evaluation mode
  with torch.inference_mode():
    test_pred=model(X_test)
    test_loss=loss_fn(test_pred,y_test)
  train_loss_history.append(train_loss)
  test_loss_history.append(test_loss)
  print(f"Epoch number{epoch+1} test loss: {test_loss.item()} train_loss: {train_loss.item()}")
```

```python
final_weight=model.weights.item()
final_bias=model.bias.item()
```
